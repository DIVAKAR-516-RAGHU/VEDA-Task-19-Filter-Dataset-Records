## Filter Dataset Records Using Pandas

**Objective:**  
Filter dataset records using one or more conditions with Pandas.

**Tools:** Python, Pandas, Jupyter Notebook

```python
import pandas as pd
```

## 1. Load the Titanic Dataset

The Titanic dataset contains passenger information such as age, gender,
passenger class, fare, and survival status.

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

## 2. Filter Passengers Based on Age

We can use a condition to retrieve passengers whose age is greater than 30.

```python
age_filtered = df[df["Age"] > 30]

age_filtered.head()
```

**Condition:** `Age > 30`

This retrieves only passengers whose age is greater than 30.

## 3. Filter Passengers Based on Gender

We can filter the dataset to retrieve only female passengers.

```python
female_passengers = df[df["Sex"] == "female"]

female_passengers.head()
```

**Condition:** `Sex == "female"`

This retrieves only the records where the passenger's gender is female.

## 4. Filter Using Multiple Conditions

Pandas allows multiple conditions using `&`.

Here, we retrieve passengers who are female and older than 30.

```python
female_above_30 = df[(df["Sex"] == "female") & (df["Age"] > 30)]

female_above_30.head()
```

**Conditions:**
- Passenger must be female
- Passenger must be older than 30

The `&` operator means both conditions must be satisfied.
Parentheses are used around each condition.

## 5. Filter Passengers Based on Passenger Class and Fare

Retrieve passengers from first class who paid a fare greater than 50.

```python
first_class_high_fare = df[
    (df["Pclass"] == 1) & (df["Fare"] > 50)
]

first_class_high_fare.head()
```

**Conditions:**
- Passenger class must be 1
- Fare must be greater than 50

The `&` operator combines both conditions.

## 6. Using OR Condition

The `|` operator is used when either of the conditions can be true.

Here, we retrieve passengers who are either in first class or second class.

```python
first_or_second_class = df[
    (df["Pclass"] == 1) | (df["Pclass"] == 2)
]

first_or_second_class.head()
```

## 7. Check the Number of Filtered Records

The `shape` attribute can be used to determine how many records satisfy
each filtering condition.

```python
print("Age > 30:", len(age_filtered))
print("Female passengers:", len(female_passengers))
print("Female and Age > 30:", len(female_above_30))
print("First class and Fare > 50:", len(first_class_high_fare))
print("First or Second class:", len(first_or_second_class))
```

## Conclusion

In this task, Pandas filtering was used to retrieve relevant records from
the Titanic dataset.

### Filtering techniques learned:
- Single-condition filtering using `>`
- Equality filtering using `==`
- Multiple conditions using `&`
- OR conditions using `|`
- Parentheses for combining multiple conditions
- Checking the number of filtered records using `len()` and `shape`

Pandas filtering helps extract specific records from large datasets
efficiently.
